```python
import numpy as np
```

```python
def himmelblau_problem(x1,x2):
    himmelblau = np.exp(
            8 - 0.01 * ((x1**2 + x2 - 11) ** 2 + (x1 + x2**2 - 7) ** 2 + x1 - 2 * x2)
        )
    butterot = np.sin(x1 + x2) + 0.4 * (x1 - x2) ** 2 - 1.5 * x1 + 2.5 * x2 + 9.0
    butterot = ((butterot / 79.4) + 1.0) * 1720
    return np.array([himmelblau, butterot]).T
```
