```python
from pathlib import Path
from paretodo.data_models import InvestigationSpace, InvestigationSpaceVariable
```

### Setup Investigation Space

```python
LOCAL_DIR = "../.local/bio_is/"
Path(LOCAL_DIR).mkdir(parents=True, exist_ok=True)

filepaths = {
    "investigation_space": f"{LOCAL_DIR}/is_2d.json",
}

investigation_space = InvestigationSpace(
    variables=[
        InvestigationSpaceVariable(
            name="temperature",
            min=30.0,
            max=40.0,
            discretization=0.001,
            expected_error=0.1,
        ),
        InvestigationSpaceVariable(
            name="ph_value",
            min=6.0,
            max=8.0,
            discretization=0.001,
            expected_error=0.1,
        ),
    ]
)
investigation_space.to_json(Path(filepaths["investigation_space"]))
```
