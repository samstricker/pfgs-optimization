### Setup Investigation Space for Himmelblau-Butterot problem

```python
import os
from pathlib import Path
from paretodo.data_models import InvestigationSpace, InvestigationSpaceVariable

FOLDER = "../.local/hb_is/"
os.makedirs(FOLDER, exist_ok=True)

investigation_space = InvestigationSpace(
    variables=[
        InvestigationSpaceVariable(
            name="initial_glucose_concentration",
            min=0.0,
            max=8.0,
            discretization=0.001,
            expected_error=0.1,
        ),
        InvestigationSpaceVariable(
            name="initial_glutamine_concentration",
            min=0.0,
            max=8.0,
            discretization=0.001,
            expected_error=0.8,
        ),
    ]
)
investigation_space.to_json(Path(f"{FOLDER}is.json"))
```
