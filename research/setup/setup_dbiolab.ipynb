## Setup of the Default Digital BioLab files
These fiels are later used as default but can be overwritten by the Simulator Class.

```python
from pathlib import Path

from dbiolab.mammalian_cellculture_simulation import (
    CellType,
    FeedConcentrations,
    FeedingEvent,
    MammalianRecipe,
    NoiseParameters,
    ReactorType,
    plot_reaction_evolution_timeseries,
)

LOCAL_DIR = "../.local/dbiolab_defaults"
Path(LOCAL_DIR).mkdir(parents=True, exist_ok=True)
filepaths = {
    "reactor": f"{LOCAL_DIR}/MTP_1.json",
    "celltype": f"{LOCAL_DIR}/celltype_paper_1.json",
    "recipe": f"{LOCAL_DIR}/recipe.json",
}
```

```python
reactor = ReactorType(
    name="MTP_1",
    volume=3.0,
    scale_factor=5.0,
    growth_inhibition=0.95,
)
reactor.to_json(filepaths["reactor"])
```

```python
celltype = CellType(
    name="celltype_paper_1",
    mu_max=0.0451639932,
    k_lysis=0.0446990515,
    k_deg_gln=0.0013441722,
    k_death_max=0.0101556537,
    k_death_mu=0.0145125692,
    k_sat_lac=141.4574131165,
    k_sat_ammonia=37.778062819,
    k_sat_glc=1.1634941712,
    k_sat_gln=0.2123961928,
    y_ammonia=9.98e-08,
    y_lactate=0.8061528449,
    y_cells_glc=153303245.24868935,
    y_cells_gln=8363679425.5848055,
    m_glc=4e-10,
    m_gln=0.0,
    ph_opt=6.7,
    ph_robustness_factor=1.1,
    e_activation=33.4,
    temperature_robustness_factor=1.1,
    y_product_from_total=9.98e-08,
    y_product_from_viable=4.9308e-06,
    hcp_per_cell=2e-6,
    k_agg_product=3e-6,
)
celltype.to_json(filepaths["celltype"])
```

```python
initial_conditions = FeedConcentrations(
    c_product=0.0,
    x_total=800000.0,
    x_viable=800000.0,
    x_dead=0.0,
    c_glc=4.0,
    c_gln=4.0,
    c_lac=0.0,
    c_ammonia=0.0,
    c_aggregates=0.0,
)
```

```python
feed_event_1 = FeedingEvent(
    time_point=120.0,
    feed_concentrations=FeedConcentrations(
        c_product=0.0,
        x_total=0.0,
        x_viable=0.0,
        x_dead=0.0,
        c_glc=2.0,
        c_gln=1.0,
        c_lac=0.0,
        c_ammonia=0.0,
        c_aggregates=0.0,
    ),
)
feed_event_2 = FeedingEvent(
    time_point=240.0,
    feed_concentrations=FeedConcentrations(
        c_product=0.0,
        x_total=0.0,
        x_viable=0.0,
        x_dead=0.0,
        c_glc=2.0,
        c_gln=1.0,
        c_lac=0.0,
        c_ammonia=0.0,
        c_aggregates=0.0,
    ),
)
feeding_strategy = [feed_event_1, feed_event_2]
```

```python
noise_parameters = NoiseParameters()
```

```python
recipe = MammalianRecipe(
    name="Test Recipe",
    celltype=celltype,
    temperature=33.4,  # range 30-40
    ph_value=7.2,  # range 6-8
    initial_conditions=initial_conditions,
    feeding_strategy=feeding_strategy,
    runtime=336.0,
    measurement_intervals=12.0,
    noise_parameters=noise_parameters,
)
recipe.to_json(filepaths["recipe"])
```
