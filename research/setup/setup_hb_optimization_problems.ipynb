```python
import os
from pathlib import Path
from paretodo.data_models import OptimizationAxis, OptimizationProblem

FOLDER = "../.local/hb_opt/"
os.makedirs(FOLDER, exist_ok=True)
```

### Setup Optimization Problem

```python
optimization_problem = OptimizationProblem(
    problem=[
        OptimizationAxis(
            var_name="product_concentration",
            minimize=False,
            optimization_type="objective",
            lower_bound=None,
            upper_bound=None,
        ),
        OptimizationAxis(
            var_name="product_concentration",
            minimize=False,
            optimization_type="standard_deviation",
            lower_bound=None,
            upper_bound=None,
        ),
    ]
)
optimization_problem.to_json(Path(f"{FOLDER}hb_opt.json"))
```
