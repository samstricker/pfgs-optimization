```python
import polars as pl

from helpers.data.dataset import Dataset
from paretodo import InvestigationSpace, InvestigationSpaceVariable,  OptimizationProblem, OptimizationAxis
```

```python
DATA_PATH = ".data"
DATASET_NAME = "example_data"
```

```python
data = pl.read_csv(f"{DATA_PATH}/{DATASET_NAME}.csv")
data
```

```python
X_COLUMNS = ['x1',
 'x2',
 'x3',
 'x4']
Y_COLUMNS = ['y1',
 'y2']
```

```python
dataset = Dataset(
    dataset_type="experiments",
    x_labels=X_COLUMNS,
    x_data=data.select(X_COLUMNS).to_numpy(),
    y_labels=Y_COLUMNS,
    y_data=data.select(Y_COLUMNS).to_numpy(),
)
dataset.to_json(f"{DATA_PATH}/{DATASET_NAME}_dataset.json")
```

```python
space = InvestigationSpace(
    variables=[
        InvestigationSpaceVariable(
            name="x1",
            min=50.0,
            max=95.0,
            discretization=5.0,
            expected_error=0.1,
        ),
        InvestigationSpaceVariable(
            name="x2",
            min=5.0,
            max=60.0,
            discretization=1.0,
            expected_error=0.1,
        ),
        InvestigationSpaceVariable(
            name="x3",
            min=2.0,
            max=5.0,
            discretization=0.5,
            expected_error=0.1,
        ),
        InvestigationSpaceVariable(
            name="x4",
            min=0.5,
            max=2.0,
            discretization=0.1,
            expected_error=0.1,
        ),
    ]
)
space.to_json(f"{DATA_PATH}/{DATASET_NAME}_investigation_space.json")   
```

```python
optimization_problem = OptimizationProblem(
    problem=[
        OptimizationAxis(
            var_name="y1",
            minimize=False,
            optimization_type="objective",
            lower_bound=None,
            upper_bound=None,
        ),
        OptimizationAxis(
            var_name="y1",
            minimize=False,
            optimization_type="standard_deviation",
            lower_bound=None,
            upper_bound=None,
        ),
        OptimizationAxis(
            var_name="y2",
            minimize=False,
            optimization_type="constraint",
            lower_bound=None,
            upper_bound=0.25,
        ),
    ]
)
optimization_problem.to_json(f"{DATA_PATH}/{DATASET_NAME}_optimization_problem.json")
```
